```python
# 1. baca dataset
df = pd.read_csv('/content/drive/MyDrive/Tugas_Data_Science_UAS/train.csv')

# kita bagi 3 kategori: Rendah, Sedang, Tinggi
bins = [0.0, 0.45, 0.55, 1.0]
labels = ['Rendah', 'Sedang', 'Tinggi']
df['RiskLevel'] = pd.cut(df['FloodProbability'], bins=bins, labels=labels)

# 3. misah fitur (X) dan target baru (y)
X = df.drop(columns=['id', 'FloodProbability', 'RiskLevel'])
y = df['RiskLevel'] # target kita berupa kategori (klasifikasi)

print("Jumlah data per kategori risiko:")
print(y.value_counts())
```
